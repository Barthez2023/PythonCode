```python
import cv2
```

```python
### Load the video file
video=cv2.VideoCapture("mountain.mp4")   ### cette instruction permet de dire a open cv de charger le fichier video

if not video.isOpened():
    print("Error:Could not open Video")
else:
    print("Video opened sucessfully")
```

```python
### Display the video
while True:
    ret,frame=video.read()  ## Cette instruction permet de lire le fichier video
    if not ret:
        print("End of video")
        break
    cv2.imshow("Video frame",frame)   ## cette instruction permet d'afficher le fichier video
    
    """Si on appui sur 'q' on arrete la lecture de la video """
    if cv2.waitKey(25) & 0xFF==ord('q'):  ## est une condition d’arrêt classique dans les programmes OpenCV qui affichent une fenêtre ou un flux vidéo.
        break


video.release() ## cette function permet de liberer la memoire utiliser par le fichier video
cv2.destroyAllWindows()  ## cette instruction permet de fermer le fenetre video
```

```python
### write a video
### Load the video file
video=cv2.VideoCapture("mountain.mp4")   ### cette instruction permet de dire a open cv de charger le fichier video

if not video.isOpened():
    print("Error:Could not open Video")

fourcc=cv2.VideoWriter_fourcc(*"mp4v")  ## cette instruction permet de definir le codec video
width=int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height=int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
out=cv2.VideoWriter("output.mp4",fourcc,30,(width,height),True)  ## cette instruction permet de creer un fichier video

while True:
    ret,frame=video.read()  ## Cette instruction permet de lire le fichier video
    if not ret:
        print("End of video")
        break
    out.write(frame)  ## cette instruction permet d'ecrire le fichier video
out.release()  ## cette instruction permet de liberer la memoire utiliser par le fichier video
video.release()  ## cette instruction permet de liberer la memoire utiliser par le fichier video
cv2.destroyAllWindows()  ## cette instruction permet de fermer le fenetre video
```

```python
### write a video
### Load the video file
video=cv2.VideoCapture("mountain.mp4")   ### cette instruction permet de dire a open cv de charger le fichier video

if not video.isOpened():
    print("Error:Could not open Video")

fourcc=cv2.VideoWriter_fourcc(*"mp4v")  ## cette instruction permet de definir le codec video
width=int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height=int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
out=cv2.VideoWriter("output_new.mp4",fourcc,30,(width,height),True)  ## cette instruction permet de creer un fichier video

while True:
    ret,frame=video.read()  ## Cette instruction permet de lire le fichier video
    if not ret:
        print("End of video")
        break
    frame=frame+20  ## cette instruction permet d'ajouter 20 a chaque pixel de l'image
    out.write(frame)  ## cette instruction permet d'ecrire le fichier video
out.release()  ## cette instruction permet de liberer la memoire utiliser par le fichier video
video.release()  ## cette instruction permet de liberer la memoire utiliser par le fichier video
cv2.destroyAllWindows()  ## cette instruction permet de fermer le fenetre video
```
