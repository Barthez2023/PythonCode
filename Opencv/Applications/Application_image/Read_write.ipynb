En OpenCV, la différence entre RGB et BGR vient uniquement de l’ordre des canaux de couleur dans l’image.
1️⃣ RGB (Red – Green – Blue)
Ordre des canaux : Rouge, puis Vert, puis Bleu.

C’est l’ordre utilisé par la plupart des bibliothèques graphiques (Pillow, Matplotlib, etc.) et par la majorité des formats d’images (PNG, JPG…).

2️⃣ BGR (Blue – Green – Red)
Ordre des canaux : Bleu, puis Vert, puis Rouge.

C’est l’ordre par défaut dans OpenCV lorsqu’on lit une image avec cv2.imread() ou qu’on capture une image avec une caméra via cv2.VideoCapture().

OpenCV a historiquement choisi cet ordre pour des raisons de compatibilité avec les anciennes API C et certaines cartes vidéo.

3️⃣ Conséquences pratiques
Si tu affiches une image BGR avec une fonction qui s’attend à du RGB (par exemple matplotlib.pyplot.imshow()), les couleurs seront inversées :

Les rouges apparaissent bleus.

Les bleus apparaissent rouges.

```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
image=cv2.imread("Images/giraffe-Kenya.png")  ## read an image and returns a numpy array
```

```python
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)  ## convert the image to RGB color space
image_hsv=cv2.cvtColor(image,cv2.COLOR_BGR2HSV)  ## convert the image to HSV color space
```

```python
plt.imshow(image_rgb)  ## display the image
```

```python
lower_white=np.array([0, 0, 190])  ## lower bound for white color in RGB
upper_white=np.array([30, 80, 255])  ## upper bound for white color in RGB


""""On veut souvent sélectionner des pixels d'une certaine couleur dans une image (exemple : détecter le blanc d'un fond ou d'un texte).
Pour cela, on définit une plage de valeurs pour chaque canal de couleur. Ici, on utilise le RGB : [R, G, B].
lower_white = np.array([0, 0, 190])

C'est la borne inférieure de la couleur blanche.
En RGB :
Rouge ≥ 0
Vert ≥ 0
Bleu ≥ 190
Donc seuls les pixels très clairs, avec une composante bleue très forte (≥190) seront considérés comme blancs.
upper_white = np.array([30, 80, 255])
C'est la borne supérieure de la couleur blanche.
En RGB :
Rouge ≤ 30
Vert ≤ 80
Bleu ≤ 255
On définit ainsi un intervalle de tolérance, pour inclure des variations de blanc légèrement jaunâtre ou grisâtre.

En pratique, on utilise ces bornes avec cv2.inRange() (OpenCV) pour créer un masque binaire où les pixels blancs deviennent 1 (ou 255) et le reste devient 0 :
import cv2

mask = cv2.inRange(image, lower_white, upper_white)
"""
```

```python
mask = cv2.inRange(image_hsv, lower_white, upper_white)  ## create a mask for white color in RGB
```

```python
plt.imshow(mask) ## display the mask
```

```python
black_hsv=np.uint8([[[120,255,0]]])  ## create a black color in HSV
black_bgr=cv2.cvtColor(black_hsv,cv2.COLOR_HSV2BGR)  ## convert the black color to BGR
```

```python
image_result=image.copy()  ## create a copy of the image to draw on it

image_result[mask>0] = black_bgr  ## set the pixels in the mask to black in the result image
```

```python
image_result_rgb=cv2.cvtColor(image_result,cv2.COLOR_BGR2RGB)  ## convert the result image to RGB color space
```

```python
## Plot the result image
fig,axs=plt.subplots(1,3,figsize=(12,5))
axs[0].imshow(image_rgb)  ## display the original image
axs[0].set_title("Original Image")  ## set title for the original image
axs[0].axis('off')  ## hide the axes for the original image

axs[1].imshow(mask)  ## display the mask
axs[1].set_title("Binary Mask")  ## set title for the mask
axs[1].axis('off')  ## hide the axes for the mask


axs[2].imshow(image_result_rgb)  ## display the result image
axs[2].set_title("OutPut Image")  ## set title for the result image
axs[2].axis('off')  ## hide the axes for the result image


plt.tight_layout()  ## adjust the layout to avoid overlap
plt.show()  ## show the plot with the original image, mask, and result image
```

```python
cv2.imwrite("original_rgb.jpg",image_rgb)
cv2.imwrite("mask.jpg",mask)
cv2.imwrite("output_rgb.jpg",image_result_rgb)
```

IMAGE RESIZING,SCALING AND INTERPOLATION

-   Image resizing:Adjusting image dimension. Ex: 640x480->4:3   1280x720 -> 16:9 
-   Image Scaling: Adjusting  image dimension but maintaining aspect ratio. Ex HD(16:9) -> Full HD(16:9)
-   Interpolation: Determines how pixel values are computed when decreasing or increasing the size of an image


1️⃣ Image Resizing (Redimensionnement)

Redimensionner une image, c’est changer sa largeur et sa hauteur.
Exemple : passer d’une image de 800×600 pixels à 400×300 pixels.
On peut réduire (downsampling) ou agrandir (upsampling) une image.

Pourquoi ?

Adapter l’image à un modèle de machine learning.
Réduire la taille pour accélérer le traitement.
Agrandir pour l’affichage ou l’édition.

2️⃣ Scaling (Mise à l’échelle)

Scaling est un type particulier de redimensionnement.
Il consiste à multiplier les dimensions de l’image par un facteur.

3️⃣ Interpolation

Quand on change la taille d’une image, on a des pixels qui disparaissent ou qui doivent être créés. L’interpolation sert à estimer la couleur de ces nouveaux pixels.

3️⃣ Interpolation

Quand on change la taille d’une image, on a des pixels qui disparaissent ou qui doivent être créés. L’interpolation sert à estimer la couleur de ces nouveaux pixels.

Types courants :
Type	            Usage	                Description
Nearest Neighbor	Rapide, simple	        Prend la couleur du pixel le plus proche. Résultat pixellisé si agrandi.
Bilinear	        Moyen	                Fait une moyenne pondérée des 4 pixels voisins. Plus lisse que nearest neighbor.
Bicubic	            Lent, haute qualité	    Utilise 16 pixels voisins pour interpoler. Résultat très lisse, souvent utilisé pour agrandir.
Lanczos	            Très haute qualité	    Utilise plusieurs pixels voisins, excellent pour réduction ou agrandissement, mais plus lent.

AFFINE AND PERSPECTIVE TRANSFORMATION

Perspective Transformation


```python
image=cv2.imread("Images/giraffe-Kenya.png")  ## read an image and returns a numpy array
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)  ## convert the image to RGB color space
plt.imshow(image_rgb)  ## display the image
```

```python
rows,colum,_ = image_rgb.shape  ## get the shape of the image
## pour effectuer la perspective transformation, on a besoin de 4 points de l'image d'origine et de 4 points de destination
points_src=np.float32([[50, 50], [colum-50, 50], [50, rows-50], [colum-50, rows-50]])  ## source points in the original image
points_dst=np.float32([[100, 100], [colum-100, 50], [100, rows-10], [colum-50, rows-100]])  ## destination points in the transformed image
""" dest_point sont les points que l'on veut obtenir apres la transformation .
Transformation matrix for perspective transformation
3x3

a,b,c
d,e,f
g,h,i

a,b,c :is responsible for horizontal scaling,rotation and translation
d,e,f :is responsible for vertical scaling,rotation and translation
g,h :is perpective compoments that account for depth and skewing effects(effets de profondeur et d'inclinaison)

"""
## on utilise cv2.getPerspectiveTransform pour obtenir la matrice de transformation
M_matrix=cv2.getPerspectiveTransform(points_src, points_dst)  ## get the transformation matrix
print(M_matrix)  ## print the transformation matrix
```

```python
"""Now M_matrix we will use for warp perspective function."""
perspective_image=cv2.warpPerspective(image_rgb, M_matrix, (colum, rows))  ## apply the perspective transformation to the image using the transformation matrix
plt.figure(figsize=(10, 5))  ## set the figure size

plt.subplot(1, 2, 1)  ## create a subplot for the original image
plt.imshow(image_rgb)  ## display the original image
plt.title("Original Image")  ## set title for the original image
plt.axis('off')  ## hide the axes for the original image

plt.subplot(1, 2, 2)  ## create a subplot for the transformed image
plt.imshow(perspective_image)  ## display the transformed image
plt.title("Perspective Transformed Image")  ## set title for the transformed image
plt.axis('off')  ## hide the axes for the transformed image
```

Applying Blur Filters:Average,Gaussian,Median

Bluring is aṡtechnique which reduces the sharp transitions in an image by averaging pixel values in a local neigborhood.

Bluring :Reducing noise ,Smoothing images(Flou : réduction du bruit, lissage des images).

Gaussian Blur uses a weighted average ,where pixels closer to the center have more influence .This produces a smoother and mor natural blur.
Average blur :each pixel in the output image is the average of the surrounding pixel values within the kernel window

Median Blur replaces each pixel with the median of the surrounding pixels ,effective at removing salt-and-pepper noise.

1. Average Blur (Moyenne)
Chaque pixel est remplacé par la moyenne arithmétique des pixels voisins dans une fenêtre de taille k × k.
Mais il a tendance à baver sur les contours (edges) et à produire une perte de netteté.


2. Gaussian Blur
Chaque pixel est remplacé par une moyenne pondérée, où les pixels proches du centre ont plus de poids que ceux plus éloignés (distribution gaussienne).
Plus naturel que le blur moyen car il préserve mieux les contours.
Utilisé pour réduire le bruit gaussien, lisser avant détection de contours (Canny, Hough, etc.).
Exemple :
Si tu floutes une image, les bords restent un peu plus clairs/nets que dans le blur moyen.

3. Median Blur
Chaque pixel est remplacé par la médiane des pixels voisins (au lieu de la moyenne).
Excellent pour supprimer le "salt-and-pepper noise" (bruit poivre et sel : pixels blancs/noirs isolés).
Préserve mieux les contours nets que Average ou Gaussian.
⚠️ Plus coûteux en calcul que Average.

👉 En résumé :

- Average Blur = simple mais floute tout.
- Gaussian Blur = plus doux, mieux pour traitement d’images.
- Median Blur = meilleur pour enlever du bruit "poivre et sel".

Edge Detection Using Sobel,Canny et Laplacian(
Détection des contours à l'aide de Sobel, Canny et Laplacian)

Egdes represents sharp changes in intessity (or color) in an image.These changes often correspond to object boundaries ,textures, or other key features.
We use edge detection for feature extraction.
We use edge detection for shape analysis.
And we use edge detection for pre-processing.

So we use edge detection to extract features.
We use edge detection for pre-processing for my CNN or AI models, and we use edge detection also for shape analysis.


- The Sobel filter dectects edges in horizontal and vertical directions by calculating gradients.Below filters mathematically mirrors the derivate operation.

- Laplacian detects edges by finding regions of rapid intensity change.Unlike Sobel ,it  combines x- and y- directons derivatives in a single operation.The Laplacian kernel below is a discrete approximation of this continuous mathematical derivate .Below filters mathematically mirrors the derivate operation.
Both kernels are performing 2nd order Derivative.
Laplacian detects sudden change in intensity ,highlighting the edge.

- Canny edge  :The canny algorithm detects edges using muti-step processing ,combining Sobel and thresholding techniques.

Step1: Convert to Grayscale because edges are present in color change information not in colors

step2: Apply Gaussian Blur which adds smoothness to image making it less sensitive to small color change variations.

step3.1 : Compute intensity Gradients using the Sobel operator approximate the first derivatives in x and y-directions.

step3.2:step3.1 calculate magnitude but we also need angle of the edge and we can get using:
O=arctang(gamma y/gamma x)

step4: Nom -Maximum Suppression ->for each pixel compare the gradient magnitude to its two neighbors along the gradient direction
->If the pixel's gradients is not the largest ,suppress(set to 0)


step5:Double Thresholding
-> High Threshold:pixels above this are strong edges=100
-> low Threshold:pixels below this are suppressed entirely=50
-> pixels between two  Thresholds are classified as weak edges.


step6:Edge tracking by Hysteris




```python
image=cv2.imread("Images/giraffe-Kenya.png",cv2.IMREAD_GRAYSCALE)  ## read an image and returns a numpy array

## Sobel edge detection
sobel_x=cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)  ## apply the Sobel filter in x direction
sobel_y=cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)  ## apply the Sobel filter in y direction

sobel_combined=cv2.magnitude(sobel_x, sobel_y)  ## calculate the magnitude of the Sobel filter


## Laplacian edge detection
laplacian=cv2.Laplacian(image, cv2.CV_64F,ksize=3)  ## apply the Laplacian filter to the image (on pouvons avoir des valeurs positives  et negatives)
laplacian_abs=cv2.convertScaleAbs(laplacian)  ## convert the Laplacian filter to absolute values (Convertir toutes les valeurs envaleurs positives)

## cany edge detection
canny_edges=cv2.Canny(image, 100, 200)  ## apply the Canny edge detection to the image with thresholds(seuils) 100 and 200


titles = ["Original", "Sobel ", "Laplacian", "Laplacian Abs","Canny Edges"]  ## titles for the images
images = [image, sobel_combined, laplacian, laplacian_abs, canny_edges]  ## list of images to display

plt.figure(figsize=(5, 5))  ## set the figure size 
for i in range(5):  ## iterate over the images
    plt.subplot(3, 2, i+1)  ## create a subplot for each image
    plt.imshow(images[i], cmap='gray')  ## display the image
    plt.title(titles[i])  ## set title for the image
    plt.axis('off')  ## hide the axes for the image
plt.tight_layout()  ## adjust the layout to avoid overlap
plt.show()  ## show the plot with the original image, Sobel, Laplacian



```

# conclusion
- Sobel:met en évidence les bords orientés, mais sensible au bruit.Rapide, simple mais génère du bruit si l’image n’est pas lissée.
- Laplacian :Utilise la dérivée seconde (2ᵉ ordre).
Détecte les zones où l’intensité change brusquement (contours + points d’inflexion).
Donne des contours plus nets, mais aussi plus sensibles au bruit que Sobel.Bords très fins, mais parfois trop de détails.Détecte contours dans toutes directions en une seule passe.Très sensible au bruit, souvent on applique un flou (GaussianBlur) avant.
- Canny:Méthode plus avancée (multi-étapes) :
Filtrage gaussien → réduction du bruit.
Calcul du gradient (comme Sobel) → intensité + direction des bords.
Non-maximum suppression → affinage des bords (garde les pixels les plus forts).
Double seuil + hystérésis → garde les vrais contours, élimine les faux.
Meilleure détection de contours, propre, avec moins de faux positifs. Robuste, précis, supprime le bruit,mais plus lent, nécessite un réglage fin des seuils.


👉 En résumé :

- Sobel → bon pour analyse simple des gradients (X, Y).
- Laplacian → met en valeur tous les contours, mais bruité.
- Canny → le plus fiable pour une détection nette et robuste.

Clache

Clache improves constrast by applying histogram equalizqtion in small regions or tiles of the image.Unlike global histogram equalization ,which enhances constrast for the entire image,clahe focuses on localized constract while avoiding artificials like noise amplification.

```python
image=cv2.imread("Images/giraffe-Kenya.png")  ## read an image and returns a numpy array

image_gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)  ## convert the image to grayscale

equalised_image=cv2.equalizeHist(image_gray)  ## apply histogram equalization to the grayscale image
clahe=cv2.createCLAHE(clipLimit=2.0, tileGridSize=(4, 4))  ## create a CLAHE object with clip limit and tile grid size
clahe_equalised_image=clahe.apply(image_gray)  ## apply CLAHE to the grayscale image


plt.figure(figsize=(10, 5))  ## set the figure size

plt.subplot(1, 3, 1)  ## create a subplot for the original image
plt.imshow(image_gray,cmap='gray')  ## display the original grayscale image 
plt.title("Original Grayscale Image")  ## set title for the original image  
plt.axis('off')  ## hide the axes for the original image

plt.subplot(1, 3, 2)  ## create a subplot for the equalized image
plt.imshow(equalised_image,cmap='gray')  ## display the equalized image
plt.title("Equalized Image")  ## set title for the equalized image
plt.axis('off')  ## hide the axes for the equalized image

plt.subplot(1, 3, 3)  ## create a subplot for the CLAHE equalized image
plt.imshow(clahe_equalised_image,cmap='gray')  ## display the CLAHE
plt.title("CLAHE Equalized Image")  ## set title for the CLAHE equalized image
plt.axis('off')  ## hide the axes for the CLAHE equalized image


plt.tight_layout()  ## adjust the layout to avoid overlap
plt.show()  ## show the plot with the original image, equalized image and CLAHE
```

```python
image=cv2.imread("Images/shapes.jpg")  ## read an image and returns a numpy array
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)  ## convert the image to RGB color space
image_gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)  ## convert the image to grayscale


plt.imshow(image_rgb)  ## display the image
```

```python
"""The next thing I need is I need a mask of this image.
I need a binary mask that I can do thresholding or I can use scanning.
So I'm going to use scanning."""

gray=cv2.Canny(image_gray, 80, 150)  ## apply the Canny edge detection to the grayscale image with thresholds(seuils) 80 and 150
plt.imshow(gray,cmap='gray')  ## display the grayscale image
```

```python
contours,hierachy=cv2.findContours(gray, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)  ## find the contours in the binary image

#plot the contours on the original image
contour_image=image_rgb.copy()  ## create a copy of the original image to draw the contours on 
cv2.drawContours(contour_image, contours, -1, (0, 255, 0), 5 )  ## draw all the contours on the image with green color and thickness of 3 pixels

plt.figure(figsize=(8, 8))  ## set the figure size
plt.subplot(1, 3, 1)  ## create a subplot for the original image
plt.imshow(image_rgb)  ## display the original image    
plt.title("Original Image")  ## set title for the original image


plt.subplot(1, 3, 2)  ## create a subplot for the image with contours
plt.imshow(gray,cmap='gray')  ## display the grayscale image
plt.title("Canny Edges")  ## set title for the grayscale image      


plt.subplot(1, 3, 3)  ## create a subplot for the image with contours
plt.imshow(contour_image)  ## display the image with the contours
plt.title("Contours")  ## set title for the image with contours
```

```python
### on peut remplacer les contour en leurs dsissnat par des rectangles
print(f"Number of contours found: {len(contours)}")  ## print the number of contours found
rectangle_image=image_rgb.copy()  ## create a copy of the original image to draw the contours
for contour in contours:  ## iterate over the contours
    x,y,w,h=cv2.boundingRect(contour)  ## get the bounding box for each contour
    cv2.rectangle(rectangle_image, (x, y), (x+w, y+h), (255, 0, 0), 3)  ## draw a rectangle around each contour with blue color and thickness of 3 pixels   


plt.figure(figsize=(8, 8))  ## set the figure size
plt.subplot(1, 3, 1)  ## create a subplot for the original image
plt.imshow(image_rgb)  ## display the original image    
plt.title("Original Image")  ## set title for the original image


plt.subplot(1, 3, 2)  ## create a subplot for the image with contours
plt.imshow(gray,cmap='gray')  ## display the grayscale image
plt.title("Canny Edges")  ## set title for the grayscale image      


plt.subplot(1, 3, 3)  ## create a subplot for the image with contours
plt.imshow(rectangle_image)  ## display the image with the contours
plt.title("Contours")  ## set title for the image with contours

```
